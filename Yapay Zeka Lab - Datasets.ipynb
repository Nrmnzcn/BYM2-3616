# Dataset 1: Differentiated Thyroid Cancer Recurrence

Link: https://archive.ics.uci.edu/dataset/915/differentiated+thyroid+cancer+recurrence

```python
# dataset
differentiated_thyroid_cancer_recurrence = fetch_ucirepo(id=915)

# data (as pandas dataframes)
X = differentiated_thyroid_cancer_recurrence.data.features
y = differentiated_thyroid_cancer_recurrence.data.targets
```

```python
X.head(8)
```

## Uyarı 1:

Verisetinde "String" ifadeler "Nümerik" değerlere çevrilmeli. Çevirmeden Makine öğrenmesi yöntemlerini uygulamaya çalıştığımız için hata veriyordu.

```python
# Kategorik sütunları seç
cat_cols = X.select_dtypes(include="object").columns

# String -> Integer dönüşümü
for col in cat_cols:
    X[col] = X[col].astype("category").cat.codes
```

```python
X.head(8)
```

## Uyarı 2:

y label değerleri de benzer şekilde Yes-No "String" ifadeler olarak verilmiş. Bunlaru da "Nümerik" değerlere yani Yes ise 1, No ise 0 olacak şekilde çevirelim.

```python
y = y.replace({"No": 0, "Yes": 1})
```

## Uyarı 3:

Bu veri setinde Y yani "Obezite Seviyesi" yukarıda görüldüğü gibi 7 sınıfa ayrılmış. Bu tip sınıflandırmalara multiclass deniyor. Ancak bu bizim seviyemize göre biraz daha zor. Bu yüzden biz bu veriyi iki sınıflıya dönüştürelim. Obez Olmayanlar:0, Obez olanlar:1
